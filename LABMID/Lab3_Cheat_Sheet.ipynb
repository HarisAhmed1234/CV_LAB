```

```

[ 0  0 0 x1 y1  1] [b ]   [y1']

[ 0  0 0 x2 y2  1] [c ]   [y2']
[x3 y3 1  0  0  0] [d ]   [x3']

In code, use `np.linalg.solve(A, B)` .

# PART C - TASK 1: THE TINY FINGERPRINT

## What the task asks

- Build a 2 x 2 scaling matrix.
- Enlarge the fingerprint by 300% in x and y.
- Keep it centered on the screen.

The task gives:

````python
sx = 3
sy = 3
````

### Approach

1. Build the 2 x 2 scale matrix.

2. Scaling around the origin would move the center, so calculate a translation that keeps the center fixed.

3. OpenCV's affine warp uses a 2 x 3 matrix.

## Core matrix

```
S = [ 3  0
      0  3 ]
```

To keep the image center fixed:

```
cx = width / 2
cy = height / 2

tx = cx - sx*cx
ty = cy - sy*cy
```

which for sx=sy=3 becomes:

```
tx = -2*cx
ty = -2*cy
```

## Complete solution

````python
import cv2
import numpy as np

# Read the fingerprint image
img = cv2.imread("fingerprint.jpg")

if img is None:
    raise FileNotFoundError("Could not find fingerprint.jpg")

# Image dimensions
h, w = img.shape[:2]

# Image center
cx = w / 2
cy = h / 2

# 300% means 3 times the original size
sx = 3.0
sy = 3.0

# The 2 x 2 scaling matrix requested by the task
S = np.array([
    [sx, 0],
    [0, sy]
], dtype=np.float32)

print("2 x 2 Scaling Matrix:")
print(S)

# Translate so scaling happens around the image center
tx = cx - sx * cx
ty = cy - sy * cy

# OpenCV warpAffine expects a 2 x 3 affine matrix.
# The final column contains translation.
M = np.array([
    [sx, 0, tx],
    [0, sy, ty]
], dtype=np.float32)

# Keep the same output canvas size
scaled = cv2.warpAffine(
    img,
    M,
    (w, h)
)

cv2.imshow("Original", img)
cv2.imshow("Scaled and Centered", scaled)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

# PART D - TASK 2: THE DIZZY SATELLITE

## What the task asks

- The image was rotated by +45 degrees.
- Formulate a 2 x 2 rotation matrix.
- Rotate it back to true north.
- Calculate new dimensions manually so corners are not cropped.

### Approach

Undo +45 degrees with:

```
theta = -45 degrees
```

## Rotation matrix

```
R(-45) =
[ cos(-45)  -sin(-45)
  sin(-45)   cos(-45) ]

     =
[ 0.7071   0.7071
 -0.7071   0.7071 ]
```

## New dimensions

For an image of width `w` and height `h` :

```
new_w = |w*cos(theta)| + |h*sin(theta)|
new_h = |h*cos(theta)| + |w*sin(theta)|
```

Use ceiling when converting to integer dimensions.

## Complete solution

````python
import cv2
import numpy as np

img = cv2.imread("satellite.jpg")

if img is None:
    raise FileNotFoundError("Could not find satellite.jpg")

h, w = img.shape[:2]

# The image was rotated +45 degrees.
# Rotate by -45 degrees to undo it.
angle = -45
theta = np.deg2rad(angle)

# 2 x 2 rotation matrix
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
], dtype=np.float32)

print("2 x 2 Rotation Matrix:")
print(R)

# Calculate the bounding-box dimensions after rotation.
# This prevents the rotated corners from being cropped.
new_w = int(np.ceil(
    abs(w * np.cos(theta)) +
    abs(h * np.sin(theta))
))

new_h = int(np.ceil(
    abs(h * np.cos(theta)) +
    abs(w * np.sin(theta))
))
# Rotate about the image center
center = (w / 2, h / 2)

M = cv2.getRotationMatrix2D(
    center,
    angle,
    1.0
)

# Recenter the rotated image inside the new canvas
M[0, 2] += new_w / 2 - center[0]
M[1, 2] += new_h / 2 - center[1]

rotated = cv2.warpAffine(
    img,
    M,
    (new_w, new_h)
)

cv2.imshow("Original", img)
cv2.imshow("Corrected Satellite", rotated)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

# PART E - TASK 3: THE FAST TRAIN BARCODE

## What the task asks

- Construct a 2 x 2 horizontal shear matrix.
- Push pixels horizontally.
- Undo the slant so the barcode becomes rectangular and readable.

The task does not provide a numerical shear coefficient `k` , so the code keeps `k` as an input.

### Approach

Forward horizontal shear:

```
H = [ 1  k
      0  1 ]
```

If the image was distorted using `+k` , undo it using:

```
H_inverse = [ 1  -k
              0   1 ]
```

## Complete solution

````python
import cv2
import numpy as np

img = cv2.imread("barcode.jpg")

if img is None:
    raise FileNotFoundError("Could not find barcode.jpg")

h, w = img.shape[:2]

# Enter the shear amount measured/given in the question.
k = float(input("Enter shear k: "))

# Inverse horizontal shear removes a rightward shear of +k.
S = np.array([
    [1, -k],
    [0,  1]
], dtype=np.float32)

print("2 x 2 Inverse Shear Matrix:")
print(S)

# Find where image corners move so we can make
# an output canvas large enough for the whole result.
corners = np.float32([
    [0, 0],
    [w, 0],
    [0, h],
    [w, h]
])

transformed_corners = corners @ S.T

min_x = transformed_corners[:, 0].min()
min_y = transformed_corners[:, 1].min()
max_x = transformed_corners[:, 0].max()
max_y = transformed_corners[:, 1].max()

# Shift the result if any coordinates become negative.
tx = -min_x
ty = -min_y

# Convert the 2 x 2 shear matrix to OpenCV's 2 x 3 format.
M = np.array([
    [1, -k, tx],
    [0,  1, ty]
], dtype=np.float32)

new_w = int(np.ceil(max_x - min_x))
new_h = int(np.ceil(max_y - min_y))

corrected = cv2.warpAffine(
    img,
    M,
    (new_w, new_h)
)
cv2.imshow("Slanted Barcode", img)
cv2.imshow("Corrected Barcode", corrected)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

## Function dictionary

`corners @ S.T`

- `@` : matrix multiplication.
- `S.T` : transpose of `S` .

`.min()`

- returns minimum element of an array.

`.max()`

- returns maximum element of an array.

`cv2.warpAffine(src, M, dsize)`

- `src` : image.
- `M` : 2 x 3 affine matrix.
- `dsize` : `(width, height)` .

# PART F - TASK 4: THE HIDDEN TREASURE MAP

## What the task asks

The map is misaligned. The X mark is hidden exactly:

```
150 pixels to the left
80 pixels up
```

The task asks you to upgrade from 2 x 2 to 3 x 3 homogeneous coordinates and translate the image back into view.

## Translation matrix

```
T =
[ 1   0  150
  0   1   80
  0   0    1 ]
```

## Complete solution

````python
import cv2
import numpy as np

img = cv2.imread("treasure_map.jpg")

if img is None:
    raise FileNotFoundError("Could not find treasure_map.jpg")

# Move the map right by 150 pixels
# and down by 80 pixels to bring the hidden mark back.
tx = 150
ty = 80

# 3 x 3 homogeneous translation matrix
T = np.array([
    [1,  0, tx],
    [0,  1, ty],
    [0,  0,  1]
], dtype=np.float32)

print("3 x 3 Translation Matrix:")
print(T)

# warpAffine needs only the first two rows.
M = T[:2, :]

h, w = img.shape[:2]

translated = cv2.warpAffine(
    img,
    M,
    (w, h)
)

cv2.imshow("Original Map", img)
cv2.imshow("Recovered Map", translated)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

# PART G - TASK 5: THE ROBOT'S ASSEMBLY LINE

## What the task asks

A microchip is rotated sideways and slightly off-center.

The task asks you to:

- combine Rotation and Translation into one 3 x 3 matrix,
- apply the combined rigid transformation,
- align the chip with the robot's gripper.

The task sheet does not provide numeric angle/translation values, so use the values given in the question/exam.

## Approach

Rotation:

```
R =
[ cos(theta)  -sin(theta)  0
  sin(theta)   cos(theta)  0
       0            0       1 ]
```

Translation:

```
T =
[ 1  0  tx
  0  1  ty
  0  0   1 ]
```

Rotate first, then translate:

```
H_rigid = T @ R
```

## Resulting matrix form

```
H_rigid =
[ cos(theta)  -sin(theta)  tx
  sin(theta)   cos(theta)  ty
      0            0        1 ]
```

## Complete solution

````python
import cv2
import numpy as np

img = cv2.imread("microchip.jpg")

if img is None:
    raise FileNotFoundError("Could not find microchip.jpg")

# Values should come from the exam/task instance.
angle_deg = float(input("Enter rotation angle (degrees): "))
tx = float(input("Enter translation in x: "))
ty = float(input("Enter translation in y: "))

theta = np.deg2rad(angle_deg)

# Homogeneous rotation matrix
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
], dtype=np.float32)
# Homogeneous translation matrix
T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
], dtype=np.float32)

# Right-to-left application:
# R happens first, then T.
H_rigid = T @ R

print("3 x 3 Rigid Transformation Matrix:")
print(H_rigid)

# warpAffine only needs the first two rows.
M = H_rigid[:2, :]

h, w = img.shape[:2]

aligned = cv2.warpAffine(
    img,
    M,
    (w, h)
)

cv2.imshow("Original Chip", img)
cv2.imshow("Aligned Chip", aligned)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

# PART H - TASK 6: THE ARCHITECT'S BLUEPRINT

## What the task asks

The blueprint is:

- too small,
- rotated randomly,
- in the wrong location.

The task asks you to combine:

```
Scale + Rotation + Translation
```

into one 3 x 3 operation.

## Approach

Use uniform scaling:

```
S =
[ s  0  0
  0  s  0
  0  0  1 ]
```

Rotation:

```
R =
[ cos(theta)  -sin(theta)  0
  sin(theta)   cos(theta)  0
      0            0       1 ]
```

Translation:

```
T =
[ 1  0  tx
  0  1  ty
  0  0   1 ]
```

Sequence:

```
scale -> rotate -> translate
```

Therefore:

```
H_similarity = T @ R @ S
```

## Complete solution

````python
import cv2
import numpy as np

img = cv2.imread("blueprint.jpg")

if img is None:
    raise FileNotFoundError("Could not find blueprint.jpg")

# Values come from the actual question/exam.
scale = float(input("Enter uniform scale factor: "))
angle_deg = float(input("Enter rotation angle (degrees): "))
tx = float(input("Enter translation in x: "))
ty = float(input("Enter translation in y: "))

theta = np.deg2rad(angle_deg)

# Uniform scaling: same factor on x and y.
S = np.array([
    [scale, 0, 0],
    [0, scale, 0],
    [0, 0, 1]
], dtype=np.float32)

# Rotation around the origin.
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
], dtype=np.float32)

# Translation.
T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
], dtype=np.float32)

# Apply in this order:
# S first -> R second -> T last.
H = T @ R @ S

print("3 x 3 Similarity Matrix:")
print(H)

M = H[:2, :]

h, w = img.shape[:2]

result = cv2.warpAffine(
    img,
    M,
    (w, h)
)

cv2.imshow("Original Blueprint", img)
cv2.imshow("Corrected Blueprint", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

## Function dictionary

- input must be radians.

`np.sin(theta)`

- input must be radians.

- `src` : image.
- `M` : 2 x 3 affine matrix.
- `dsize` : `(width, height)` .

# PART I - TASK 7: THE GLITCHED IMAGE

## What the task asks

Three landmarks have known:

```
starting coordinates
ending coordinates
```

The image was squished, slanted, rotated, and moved.

You must solve the six unknown parameters of a general affine matrix:

```
a, b, tx, c, d, ty
```

## Core equations

```
x' = a*x + b*y + tx
y' = c*x + d*y + ty
```

For each point:

```
[x  y  1  0  0  0] [a ] = [x']
[0  0  0  x  y  1] [b ]   [y']
```

Three point pairs give six equations.

## Complete solution

````python
import cv2
import numpy as np

img = cv2.imread("glitched.jpg")

if img is None:
    raise FileNotFoundError("Could not find glitched.jpg")

# Replace these with the 3 source landmarks
# and their 3 corresponding destination landmarks
# given by the exam question.
src = np.float32([
    [x1, y1],
    [x2, y2],
    [x3, y3]
])

dst = np.float32([
    [xp1, yp1],
    [xp2, yp2],
    [xp3, yp3]
])

A = []
B = []

# Each point pair contributes TWO equations:
# one for x' and one for y'.
for (x, y), (xp, yp) in zip(src, dst):

    # x' = a*x + b*y + tx
    A.append([x, y, 1, 0, 0, 0])
    B.append(xp)

    # y' = c*x + d*y + ty
    A.append([0, 0, 0, x, y, 1])
    B.append(yp)

A = np.array(A, dtype=np.float64)
B = np.array(B, dtype=np.float64)

# Solve A * parameters = B.
# Do NOT need to calculate the inverse manually.
params = np.linalg.solve(A, B)

a, b, tx, c, d, ty = params

print("Solved parameters:")
print("a =", a)
print("b =", b)
print("tx =", tx)
print("c =", c)
print("d =", d)
print("ty =", ty)

# Construct the resulting 3 x 3 affine matrix.
H = np.array([
    [a, b, tx],
    [c, d, ty],
    [0, 0, 1]
], dtype=np.float64)

print("Affine Matrix:")
print(H)
# OpenCV affine warp uses the top 2 rows.
M = H[:2, :].astype(np.float32)

h, w = img.shape[:2]

corrected = cv2.warpAffine(
    img,
    M,
    (w, h)
)

cv2.imshow("Glitched Image", img)
cv2.imshow("Corrected Image", corrected)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

**Important:** `x1` , `y1` , etc. above are placeholders because the task sheet describes three

correspondences but does not print numerical coordinates on the page.

# PART K - TASK 9: THE STADIUM PANORAMA

## What the task asks

Two cameras took overlapping photos of the same football field.

You must:

1. choose 4 matching points in both images,

2. calculate the homography,

3. project camera 2 into camera 1's coordinate space,

4. stitch them into one panorama.

## Approach

If:

```
pts2 = points in camera/image 2
pts1 = corresponding points in camera/image 1
```

then compute:

````python
H, mask = cv2.findHomography(pts2, pts1, 0)
````

This mapping sends:

```
image 2 -> image 1
```

Then warp image 2.

## Complete solution

````python
import cv2
import numpy as np

img1 = cv2.imread("stadium1.jpg")
img2 = cv2.imread("stadium2.jpg")

if img1 is None or img2 is None:
    raise FileNotFoundError("Could not load one or both stadium images")

# Four matching points.
# pts2 and pts1 MUST correspond in the same order.
# Replace the placeholders with the actual points.
pts2 = np.float32([
    [x1_2, y1_2],
    [x2_2, y2_2],
    [x3_2, y3_2],
    [x4_2, y4_2]
])

pts1 = np.float32([
    [x1_1, y1_1],
    [x2_1, y2_1],
    [x3_1, y3_1],
    [x4_1, y4_1]
])

# Find H that maps image 2 coordinates into image 1 coordinates.
H, mask = cv2.findHomography(
    pts2,
    pts1,
    0
)

print("Homography Matrix:")
print(H)

h1, w1 = img1.shape[:2]
h2, w2 = img2.shape[:2]

# Four corners of image 2
corners2 = np.float32([
    [0, 0],
    [w2, 0],
    [w2, h2],
    [0, h2]
]).reshape(-1, 1, 2)

# Where those corners land after H
warped_corners2 = cv2.perspectiveTransform(
    corners2,
    H
).reshape(-1, 2)

# Image 1 corners in its own coordinate system
corners1 = np.float32([
    [0, 0],
    [w1, 0],
    [w1, h1],
    [0, h1]
])

# Find complete panorama bounds.
all_corners = np.vstack([
    corners1,
    warped_corners2
])

min_x = int(np.floor(all_corners[:, 0].min()))
min_y = int(np.floor(all_corners[:, 1].min()))

max_x = int(np.ceil(all_corners[:, 0].max()))
max_y = int(np.ceil(all_corners[:, 1].max()))

panorama_w = max_x - min_x
panorama_h = max_y - min_y

# Translation moves negative coordinates into the visible canvas.
T = np.array([
    [1, 0, -min_x],
    [0, 1, -min_y],
    [0, 0, 1]
], dtype=np.float64)

# Apply H first, then translate into panorama canvas.
H_panorama = T @ H

# Warp image 2.
warped2 = cv2.warpPerspective(
    img2,
    H_panorama,
    (panorama_w, panorama_h)
)

# Create canvas and place image 1.
panorama = np.zeros_like(warped2)

x_offset = -min_x
y_offset = -min_y

panorama[
    y_offset:y_offset + h1,
    x_offset:x_offset + w1
] = img1

# Create masks for blending.
mask1 = np.zeros(
    (panorama_h, panorama_w),
    dtype=np.uint8
)

mask1[
    y_offset:y_offset + h1,
    x_offset:x_offset + w1
] = 255

mask2_original = np.ones(
    (h2, w2),
    dtype=np.uint8
) * 255

mask2 = cv2.warpPerspective(
    mask2_original,
    H_panorama,
    (panorama_w, panorama_h)
)

# Blend overlapping areas by averaging.
p1 = panorama.astype(np.float32)
p2 = warped2.astype(np.float32)

m1 = mask1 > 0
m2 = mask2 > 0

overlap = m1 & m2
p1[overlap] = (p1[overlap] + p2[overlap]) / 2

only2 = (~m1) & m2
p1[only2] = p2[only2]

panorama = np.clip(
    p1,
    0,
    255
).astype(np.uint8)

cv2.imshow("Stadium Panorama", panorama)

cv2.waitKey(0)
cv2.destroyAllWindows()
````

# PART L - TASK 10: THE MASTER FORGER

## What the task asks

Use the entire hierarchy:

1. **Linear Scale** to shrink the painting.

2. **Rigid Transformation** to rotate and move it near the frame.

3. **Projective Transformation** to map the painting's four corners into the perspective of the angled frame.

## Main mathematical idea

Scale:

```
S
```

Rigid:

```
T @ R
```

Affine/similarity stage:

```
H_affine = T @ R @ S
```

Then use the transformed four painting corners and the four frame corners to calculate

projective correction:

```
P = getPerspectiveTransform(intermediate_corners, frame_corners)
```

The complete transformation is:

```
H_total = P @ H_affine
```

Read the order right-to-left:

```
S -> R -> T -> P
```

## Complete solution

````python
import cv2
import numpy as np

painting = cv2.imread("painting.jpg")
frame = cv2.imread("frame.jpg")

if painting is None:
    raise FileNotFoundError("Could not find painting.jpg")

if frame is None:
    raise FileNotFoundError("Could not find frame.jpg")

# These values must come from the actual question/exam instance.
scale = float(input("Enter scale factor: "))
angle_deg = float(input("Enter rotation angle (degrees): "))
tx = float(input("Enter translation in x: "))
ty = float(input("Enter translation in y: "))

theta = np.deg2rad(angle_deg)

# ---------------------------------------------------------
# 1. LINEAR SCALE
# ---------------------------------------------------------
S = np.array([
    [scale, 0, 0],
    [0, scale, 0],
    [0, 0, 1]
], dtype=np.float64)

# ---------------------------------------------------------
# 2. ROTATION
# ---------------------------------------------------------
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
], dtype=np.float64)

# ---------------------------------------------------------
# 3. TRANSLATION
# ---------------------------------------------------------
T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
], dtype=np.float64)
# Scale -> rotate -> translate
H_affine = T @ R @ S

# Four corners of the original painting
hp, wp = painting.shape[:2]

painting_corners = np.float32([
    [0, 0],
    [wp - 1, 0],
    [wp - 1, hp - 1],
    [0, hp - 1]
]).reshape(-1, 1, 2)

# Find where the corners land after scale + rotation + translation.
intermediate_corners = cv2.perspectiveTransform(
    painting_corners,
    H_affine
).reshape(-1, 2)

# ---------------------------------------------------------
# 4. FOUR CORNERS OF THE ANGLED FRAME
#
# Replace the placeholders with the actual frame corners.
# Keep the same corner order as the painting corners.
# ---------------------------------------------------------
frame_corners = np.float32([
    [fx1, fy1],  # top-left
    [fx2, fy2],  # top-right
    [fx3, fy3],  # bottom-right
    [fx4, fy4]   # bottom-left
])

# Perspective matrix maps the intermediate painting corners
# exactly onto the frame corners.
P = cv2.getPerspectiveTransform(
    intermediate_corners.astype(np.float32),
    frame_corners
)

print("Projective Matrix:")
print(P)

# ---------------------------------------------------------
# 5. COMPLETE TRANSFORMATION
#
# Original painting:
# Scale -> Rotate -> Translate -> Perspective
# ---------------------------------------------------------
H_total = P @ H_affine

print("Complete 3 x 3 Matrix:")
print(H_total)

hf, wf = frame.shape[:2]

# Apply complete projective transformation in one warp.
warped_painting = cv2.warpPerspective(
    painting,
    H_total,
    (wf, hf)
)

# Create a mask to know where the painting exists.
mask = np.ones(
    (hp, wp),
    dtype=np.uint8
) * 255

warped_mask = cv2.warpPerspective(
    mask,
    H_total,
    (wf, hf)
)

# Composite warped painting onto the frame.
result = frame.copy()

result[warped_mask > 0] = warped_painting[warped_mask > 0]

cv2.imshow("Original Frame", frame)
cv2.imshow("Painting in Frame", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
````
